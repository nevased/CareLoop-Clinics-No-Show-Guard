```python
import pandas as pd

# Load cleaned appointments data
appointments = pd.read_csv("C:\\Users\\Dipali\\CareLopp_Clinic_Project\\appointments_clean.csv")

# Load enriched clinics data
clinics = pd.read_csv("C:\\Users\\Dipali\\CareLopp_Clinic_Project\\clinics_enriched.csv")

# Display appointment columns
print(appointments.columns)

# Display clinic columns
print(clinics.columns)

# Display first five appointment records
print(appointments.head())

# Display first five clinic records
print(clinics.head())

# Group appointments clinic-wise
clinic_summary = (
    appointments
    .groupby("clinic_id")
    .agg(
        appointments=("appointment_id", "count"),
        no_shows=("no_show", "sum")
    )
    .reset_index()
)

# Display the result
print(clinic_summary)


# Merge clinic summary with clinic details
clinic_summary = clinic_summary.merge(
    clinics[["clinic_id", "clinic_name", "city"]],
    on="clinic_id",
    how="right"
)

# Replace missing appointment counts with zero
clinic_summary[["appointments", "no_shows"]] = (
    clinic_summary[["appointments", "no_shows"]].fillna(0)
)

# Display the result
print(clinic_summary)

# Calculate no-show percentage
clinic_summary["no_show_pct"] = (
    100.0 * clinic_summary["no_shows"]
    / clinic_summary["appointments"].replace(0, float("nan"))
)

# Replace undefined percentages for zero-appointment clinics with 0
clinic_summary["no_show_pct"] = (
    clinic_summary["no_show_pct"].fillna(0).round(1)
)

# Sort by no-show percentage in descending order
clinic_summary = clinic_summary.sort_values(
    by="no_show_pct",
    ascending=False
)

# Display final result
print(clinic_summary)

# Calculate no-show percentage
clinic_summary["no_show_pct"] = (
    100.0 * clinic_summary["no_shows"]
    / clinic_summary["appointments"].replace(0, float("nan"))
)

# Replace undefined percentages for zero-appointment clinics with 0
clinic_summary["no_show_pct"] = (
    clinic_summary["no_show_pct"].fillna(0).round(1)
)

# Sort by no-show percentage in descending order
clinic_summary = clinic_summary.sort_values(
    by="no_show_pct",
    ascending=False
)

# Display final result
print(clinic_summary)


# Keep clinics with at least 95 appointments
clinic_summary_95 = clinic_summary[
    clinic_summary["appointments"] >= 95
]

# Display the result
print(clinic_summary_95)

# Find the clinic with the highest no-show percentage
highest_clinic = clinic_summary.iloc[0]

print("Clinic with the highest no-show rate:")
print("Clinic Name:", highest_clinic["clinic_name"])
print("City:", highest_clinic["city"])
print("Appointments:", int(highest_clinic["appointments"]))
print("No-shows:", int(highest_clinic["no_shows"]))
print("No-show Percentage:", highest_clinic["no_show_pct"], "%")
```
