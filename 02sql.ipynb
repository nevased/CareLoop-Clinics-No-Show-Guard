```python
# STEP 1: Install required library
%pip install nbformat

# STEP 2: Import libraries
import os
import nbformat as nbf

# STEP 3: Specify your SQL file name
sql_file = "C:\\Users\\Dipali\\CareLopp_Clinic_Project\\SQL scirpt.sql"

# STEP 4: Check whether the SQL file exists
if not os.path.exists(sql_file):
    raise FileNotFoundError(
        f"SQL file not found: {sql_file}\n"
        f"Current folder: {os.getcwd()}\n"
        "Place your SQL file in this folder or provide its full path."
    )

# STEP 5: Read the SQL script
with open(sql_file, "r", encoding="utf-8-sig") as file:
    sql_code = file.read()

if not sql_code.strip():
    raise ValueError("The SQL file is empty.")

# STEP 6: Create a new Jupyter Notebook
notebook = nbf.v4.new_notebook()

# STEP 7: Add notebook cells
notebook.cells = [
    nbf.v4.new_markdown_cell(
        "# SQL Script Converted to Jupyter Notebook\n"
        "This notebook contains SQL queries imported from the original SQL file."
    ),
    nbf.v4.new_markdown_cell("## Original SQL Queries"),
    nbf.v4.new_code_cell(
        "# SQL script imported as text.\n"
        "# Execute these queries using your database tool or SQL connection.\n\n"
        'sql_script = """\n' + sql_code + '\n"""'
    )
]

# STEP 8: Set notebook metadata
notebook.metadata = {
    "kernelspec": {
        "display_name": "Python 3",
        "language": "python",
        "name": "python3"
    },
    "language_info": {
        "name": "python"
    }
}

# STEP 9: Save as IPYNB
output_file = os.path.splitext(sql_file)[0] + ".ipynb"

nbf.write(notebook, output_file)

# STEP 10: Display success message
print("SQL to IPYNB conversion completed successfully!")
print("Original SQL file:", sql_file)
print("Converted notebook:", output_file)
print("SQL file size:", len(sql_code), "characters")
print("Notebook saved in:", os.path.abspath(output_file))
```
